```python
class SimpleTokenizerV1:
    def __init__(self,vocab):
        self.str_to_int=vocab
        self.int_to_str={i:s for s,i in vocab.items()}

    def encode(self,text):
        preprocessed = re.split(r'([,.:;?_!"()\']|--|\s)',text)
        preprocessed=[item.strip() for item in preprocessed if item.strip()]
        ids=[self.str_to_int[token] for token in preprocessed]
        return ids
    
    def decode(self,ids):
        text=" ".join([self.int_to_str[i]for i in ids])
        #replace space before punctuation - becuase we added space while tokenization so if we dont do this punctuation will have space before it like a "hello , world !" instead of "hello, world!"
        text=re.sub(r'\s+([,.:;?_!"()\'])', r'\1', text)
        return text
```

```python
#now we crete  a instance of the tokenizer class
tokenizer=SimpleTokenizerV1(vocab)
text="""It's the last he painted, you know," Mrs. Gisburn said with pardonable pride."""
ids=tokenizer.encode(text)
print(ids)
```

```python
class SimpleTokenizerV2:
    def __init__(self,vocab):
        self.str_to_int=vocab
        self.int_to_str={i:s for s,i in vocab.items()}

    def encode(self,text):
        preprocessed=re.split(r'([,.;:?_!"()\']|--|\s)',text)
        preprocessed=[item.strip() for item in preprocessed if item.strip()]
        preprocessed=[item if item in self.str_to_int else "<|unk|>" for item in preprocessed ]

        ids=[self.str_to_int[s] for s in preprocessed] 
        return ids
    
    def decode(self,ids):
        text=" ".join([self.int_to_str[i] for i in ids])

        text=re.sub(r'\s+([,.:;?_!"()\'])',r'\1',text)
        return text
```

```python
tokenizer=SimpleTokenizerV2(vocab)
text1="Hello, do you like tea?"
text2="In this sunlit terraces of the palace."

text=" <|endoftext|> ".join((text1,text2))

print(text)
```
