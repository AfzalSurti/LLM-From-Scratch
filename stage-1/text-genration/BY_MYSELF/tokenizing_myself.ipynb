```python
with open("Operating_System.txt","r",encoding="utf-8") as f1:
    raw_text1=f1.read()
    print("total number of characters in Operating System:",len(raw_text1))
    print(raw_text1[:99])

with open("Computer_Network.txt","r",encoding="utf-8") as f2:
    raw_text2=f2.read()
    print("total number of characters in Computer Network:",len(raw_text2))
    print(raw_text2[:99])
```

```python
import re

result1=re.split(r'([,.?;:!"()\']|--|\s)',raw_text1)
result2=re.split(r'([,.?;:!"()\']|--|\s)',raw_text2)
result1=[item.strip() for item in result1 if item.strip()]
result2=[item.strip() for item in result2 if item.strip()]
```

```python
all_words1=sorted(set(result1))
all_words2=sorted(set(result2))
print(len(all_words1))
print(len(all_words2))
```

```python
vocab1={token:integer for integer,token in enumerate(all_words1)}
vocab2={token:integer for integer,token in enumerate(all_words2)}
```

```python
for i,item in enumerate(vocab1.items()):
    if i<10:
        print(item)
print("-----")
for i,item in enumerate(vocab2.items()):
    if i<10:
        print(item)
```

```python
all_tokens1=sorted(list(set(result1)))
all_tokens2=sorted(list(set(result2)))
all_tokens1.extend(["<|unk|>","<|endoftext|>"])
all_tokens2.extend(["<|unk|>","<|endoftext|>"])
vocab1={token:integer for integer,token in enumerate(all_tokens1)}
vocab2={token:integer for integer,token in enumerate(all_tokens2)}
```

```python
class SimpleTokenizerV2:
    def __init__(self,vocab):
        self.str_to_int=vocab
        self.int_to_str={i:s for s,i in vocab.items()}

    def encode(self,text):
        preprocessed=re.split(r'([,.;:?_!"()\']|--|\s)',text)
        preprocessed=[item.strip() for item in preprocessed if item.strip()]
        preprocessed=[item if item in self.str_to_int else "<|unk|>" for item in preprocessed ]

        ids=[self.str_to_int[s] for s in preprocessed] 
        return ids
    
    def decode(self,ids):
        text=" ".join([self.int_to_str[i] for i in ids])

        text=re.sub(r'\s+([,.:;?_!"()\'])',r'\1',text)
        return text
```

```python
all_tokens = sorted(set(result1 + ["<|endoftext|>"] + result2))
all_tokens.extend(["<|unk|>"])
vocab = {token: integer for integer, token in enumerate(all_tokens)}
tokenizer = SimpleTokenizerV2(vocab)
test_string = "Concepts  TENTH  EDITION  "
encoded = tokenizer.encode(test_string)
print("Encoded:", encoded)
decoded = tokenizer.decode(encoded)
print("Decoded:", decoded)
```
