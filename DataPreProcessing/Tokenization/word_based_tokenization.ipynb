```python
with open("the-verdict.txt","r",encoding="utf-8") as f :
    raw_text=f.read()

print("total number of characterd:",len(raw_text))
print(raw_text[:99])
```

```python
# now we have to crte the tokens for all individual words so for that we used the reguler expression library of the python
import re
preprocessed = re.split(r'([,.:;?_!"()\']|--|\s)',raw_text)
preprocessed=[item.strip() for item in preprocessed if item.strip()]
print(preprocessed[:30])
```
